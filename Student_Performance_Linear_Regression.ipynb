# Assignment 1: Student Performance Prediction

## Objective
Apply **Linear Regression** to predict the student's **Exam Score** using the remaining **numeric columns** in the Student Performance Dataset.

> **Dataset note:** The provided CSV uses `exam_score` (lowercase) as the target column. The assignment statement writes `Exam_Score`; the code uses the actual column name from the provided dataset.


## 1. Import Libraries
We use pandas for data handling, NumPy for numerical calculations, scikit-learn for train/test splitting, Linear Regression, and model evaluation.


```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

## 2. Load the Dataset


```python
# Load the provided dataset.
df = pd.read_csv("student_habits_performance.csv")

# Display the first five rows.
df.head()
```

## 3. Understand the Dataset


```python
# Check dataset dimensions, column names, data types, and missing values.
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())
```

## 4. Select Numeric Features and Target

The assignment specifically asks to use the **numeric columns** as features. Therefore, categorical columns such as gender, diet quality, internet quality, etc. are not used.

`student_id` is an object/string column, so it is automatically excluded.

Target: `exam_score`


```python
# Define the target column.
target = "exam_score"

# Select all numeric columns.
numeric_features = df.select_dtypes(include=np.number).columns.tolist()

# Remove the target from the feature list.
numeric_features.remove(target)

X = df[numeric_features]
y = df[target]

print("Numeric features used:")
print(numeric_features)

print("\nFeature matrix shape:", X.shape)
print("Target shape:", y.shape)
```

## 5. Split Data into Training and Testing Sets

We use 80% of the data for training and 20% for testing. `random_state=42` makes the split reproducible.


```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

## 6. Train the Linear Regression Model


```python
# Create the Linear Regression model.
model = LinearRegression()

# Train the model using the training data.
model.fit(X_train, y_train)

print("Linear Regression model trained successfully.")
```

## 7. Make Predictions


```python
# Predict Exam Scores for the test set.
y_pred = model.predict(X_test)

# Display a few predictions.
prediction_results = pd.DataFrame({
    "Actual_Exam_Score": y_test.values,
    "Predicted_Exam_Score": y_pred
})

prediction_results.head(10)
```

## 8. Model Evaluation

The assignment requires four evaluation metrics:

- **MAE:** Mean Absolute Error
- **MSE:** Mean Squared Error
- **RMSE:** Root Mean Squared Error
- **R² Score:** Proportion of target variance explained by the model


```python
# Calculate evaluation metrics.
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print(f"MAE  : {mae:.4f}")
print(f"MSE  : {mse:.4f}")
print(f"RMSE : {rmse:.4f}")
print(f"R²   : {r2:.4f}")
```

## 9. Model Coefficients

The coefficients show how the predicted Exam Score changes, on average, for a one-unit increase in each numeric feature while the other features remain constant.


```python
coefficients = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": model.coef_
}).sort_values("Coefficient", ascending=False)

coefficients
```

## 10. Prediction Results

The following table shows actual versus predicted Exam Scores for the test set.


```python
prediction_results["Absolute_Error"] = abs(
    prediction_results["Actual_Exam_Score"] -
    prediction_results["Predicted_Exam_Score"]
)

prediction_results
```

## 11. Example: Predict a New Student's Exam Score

Below is an example prediction using the same numeric feature order used during training.


```python
# Example student input.
new_student = pd.DataFrame([{
    "age": 20,
    "study_hours_per_day": 5.0,
    "social_media_hours": 2.0,
    "netflix_hours": 1.0,
    "attendance_percentage": 90.0,
    "sleep_hours": 7.0,
    "exercise_frequency": 3,
    "mental_health_rating": 7
}])

new_prediction = model.predict(new_student)[0]

print(f"Predicted Exam Score: {new_prediction:.2f}")
```

## 12. Conclusion

A Linear Regression model was trained using the numeric features of the Student Performance Dataset to predict `exam_score`.

The model was evaluated using MAE, MSE, RMSE, and R² Score. The prediction results above compare the model's predicted scores with the actual scores from the unseen test data.
